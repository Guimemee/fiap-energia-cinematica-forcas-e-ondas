### 1. Cinemática: Movimento Retilíneo Uniformemente Variado (MRUV)
**Problema:** Um veículo trafega a $v_0 = 108\text{ km/h} = 30\text{ m/s}$ e aciona os freios desacelerando a $a = -5,0\text{ m/s}^2$. Calcular a distância de parada e o tempo até o repouso.

```python
import numpy as np
import matplotlib.pyplot as plt

v0 = 30.0 # m/s
a = -5.0 # m/s2
v_final = 0.0

# Torricelli: v^2 = v0^2 + 2*a*delta_s
delta_s = (v_final**2 - v0**2) / (2 * a)
t_parada = (v_final - v0) / a

print(f'Velocidade Inicial: {v0*3.6:.1f} km/h ({v0:.1f} m/s)')
print(f'Tempo de Frenagem:  {t_parada:.2f} s')
print(f'Distância Percorrida: {delta_s:.2f} m')

# Curva de Posição e Velocidade
t = np.linspace(0, t_parada, 100)
s = v0 * t + 0.5 * a * (t**2)
v = v0 + a * t

fig, ax1 = plt.subplots(figsize=(8.5, 4.2), dpi=300)
ax1.plot(t, s, color='#0969da', lw=2.5, label='Posição s(t) [m]')
ax1.set_xlabel('Tempo (s)', fontweight='bold')
ax1.set_ylabel('Distância Percorrida (m)', color='#0969da', fontweight='bold')
ax1.grid(True, linestyle='--', alpha=0.5)

ax2 = ax1.twinx()
ax2.plot(t, v, color='#cf222e', lw=2, linestyle='--', label='Velocidade v(t) [m/s]')
ax2.set_ylabel('Velocidade (m/s)', color='#cf222e', fontweight='bold')

plt.title('Dinâmica de Frenagem de Emergência no MRUV', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 2. Dinâmica: Plano Inclinado com Atrito
**Problema:** Bloco de massa $m = 10\text{ kg}$ sobre plano com inclinação $\theta = 30^\circ$ e coeficiente de atrito cinético $\mu = 0,20$. Determinar a aceleração de descida.

```python
m = 10.0
g = 9.81
theta_deg = 30.0
theta_rad = np.radians(theta_deg)
mu = 0.20

P_x = m * g * np.sin(theta_rad)
N = m * g * np.cos(theta_rad)
F_atrito = mu * N
F_resultante = P_x - F_atrito
aceleracao = F_resultante / m

print(f'Força Peso Tangencial Px: {P_x:.2f} N')
print(f'Força Normal N:           {N:.2f} N')
print(f'Força de Atrito Fat:      {F_atrito:.2f} N')
print(f'Força Resultante Fres:    {F_resultante:.2f} N')
print(f'Aceleração de Descida a:  {aceleracao:.2f} m/s²')
```
