# Checkpoint 6: Balanço Energético e Dimensionamento de Motor Elétrico
Cálculo de trabalho útil, perdas por rendimento e potência nominal consumida.


```python
massa = 800.0   # kg
altura = 15.0   # m
tempo = 20.0    # s
eta = 0.75      # rendimento 75%
g = 9.81

w_util = massa * g * altura
p_util_w = w_util / tempo
p_elec_w = p_util_w / eta
p_elec_cv = p_elec_w / 735.5

print(f'Trabalho Mecânico Útil:  {w_util/1000:.2f} kJ')
print(f'Potência Mecânica Útil:  {p_util_w/1000:.3f} kW')
print(f'Potência Elétrica Rede:  {p_elec_w/1000:.3f} kW')
print(f'Potência do Motor:       {p_elec_cv:.2f} cv')

```

```python
import matplotlib.pyplot as plt

categorias = ['Potência Útil (Mecânica)', 'Perdas Térmicas / Atrito', 'Potência Total (Elétrica)']
valores = [p_util_w / 1000.0, (p_elec_w - p_util_w) / 1000.0, p_elec_w / 1000.0]
cores = ['#2ca02c', '#d62728', '#1f77b4']

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(categorias, valores, color=cores, width=0.45)
ax.set_ylabel('Potência (kW)', fontweight='bold')
ax.set_title('Fluxo de Potência e Eficiência Energética no Sistema Elevador (η = 75%)', fontweight='bold')
for b in bars:
    h = b.get_height()
    ax.text(b.get_x() + b.get_width()/2.0, h + 0.15, f'{h:.2f} kW', ha='center', fontweight='bold')
ax.set_ylim(0, 9.5)
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
