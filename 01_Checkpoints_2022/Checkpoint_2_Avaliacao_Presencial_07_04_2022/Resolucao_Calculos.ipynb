# Checkpoint 2: Estatística Experimental e Propagação de Incertezas
Cálculo de média, desvio padrão amostral e intervalo de confiança.


```python
import numpy as np

amostras = np.array([25.10, 25.15, 25.10, 25.20, 25.15])
n = len(amostras)
media = np.mean(amostras)
desv_pad = np.std(amostras, ddof=1)
incerteza_media = desv_pad / np.sqrt(n)

print(f'Amostras coletadas:   {amostras}')
print(f'Média Aritmética:     {media:.3f} mm')
print(f'Desvio Padrão (s):    {desv_pad:.4f} mm')
print(f'Incerteza da Média:   {incerteza_media:.4f} mm')
print(f'Resultado Formal:     D = ({media:.2f} ± {incerteza_media:.2f}) mm')

```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
ax.errorbar([1], [media], yerr=[incerteza_media], fmt='o', color='blue', ecolor='red', elinewidth=2.5, capsize=8, label='Média ± Incerteza')
ax.scatter(np.ones_like(amostras), amostras, color='gray', alpha=0.7, label='Leituras Individuais')
ax.axhline(media, color='blue', linestyle=':', alpha=0.5)

ax.set_xlim(0.5, 1.5)
ax.set_xticks([1])
ax.set_xticklabels(['Diâmetro do Cilindro'])
ax.set_ylabel('Dimensão Medida (mm)', fontweight='bold')
ax.set_title('Dispersão Experimental e Incerteza da Média', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()

```
