# Checkpoint 4: Cinemática Vetorial de Lançamento Oblíquo
Simulação da trajetória parabólica 2D de projétil com gravidade.


```python
import numpy as np

v0 = 40.0
theta = np.radians(45.0)
g = 9.81

v0x = v0 * np.cos(theta)
v0y = v0 * np.sin(theta)

t_voo = 2 * v0y / g
h_max = (v0y**2) / (2 * g)
alcance = v0x * t_voo

print(f'Tempo de voo:     {t_voo:.3f} s')
print(f'Altura máxima:    {h_max:.2f} m')
print(f'Alcance total:    {alcance:.2f} m')

```

```python
import matplotlib.pyplot as plt
import numpy as np

t_vals = np.linspace(0, t_voo, 200)
x_vals = v0x * t_vals
y_vals = v0y * t_vals - 0.5 * g * t_vals**2

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(x_vals, y_vals, 'b-', lw=2.5, label='Trajetória Balística (θ = 45°)')
ax.scatter([alcance/2], [h_max], color='red', s=90, zorder=5, label=f'Ápice (Hmax = {h_max:.1f} m)')
ax.scatter([alcance], [0], color='green', s=90, zorder=5, label=f'Impacto (Alcance = {alcance:.1f} m)')

ax.set_xlabel('Distância Horizontal x (m)', fontweight='bold')
ax.set_ylabel('Altura Vertical y (m)', fontweight='bold')
ax.set_title('Lançamento Oblíquo: Trajetória Parabólica no Vácuo', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()

```
