# Checkpoint 3: Equilíbrio Estático e Decomposição de Forças
Cálculo matricial das forças de tração nos cabos de sustentação.


```python
import numpy as np

# Sistema linear:
# [ -cos(30°)   cos(60°) ] [ TA ] = [    0 ]
# [  sin(30°)   sin(60°) ] [ TB ]   [ 1200 ]
rad30 = np.radians(30)
rad60 = np.radians(60)

A_mat = np.array([
    [-np.cos(rad30),  np.cos(rad60)],
    [ np.sin(rad30),  np.sin(rad60)]
])
B_vec = np.array([0.0, 1200.0])

T_sol = np.linalg.solve(A_mat, B_vec)
T_A, T_B = T_sol[0], T_sol[1]

print(f'Tração no Cabo A (30°): TA = {T_A:.2f} N')
print(f'Tração no Cabo B (60°): TB = {T_B:.2f} N')
print(f'Carga Suspensa:         P  = 1200.00 N')

```

```python
import matplotlib.pyplot as plt

cabos = ['Cabo A (30°)', 'Cabo B (60°)', 'Carga P']
forcas = [T_A, T_B, 1200.0]

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(cabos, forcas, color=['#1f77b4', '#ff7f0e', '#d62728'], width=0.5)
ax.set_ylabel('Intensidade de Força (Newtons)', fontweight='bold')
ax.set_title('Equilíbrio Estático de Ponto Material em Cabos Concorrentes', fontweight='bold')
for b in bars:
    h = b.get_height()
    ax.text(b.get_x() + b.get_width()/2.0, h + 25, f'{h:.1f} N', ha='center', fontweight='bold')
ax.set_ylim(0, 1400)
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
