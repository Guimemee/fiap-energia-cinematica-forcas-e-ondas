# Checkpoint 1: Análise Cinemática de Frenagem de Veículo Autônomo
Cálculo de Torricelli e simulação temporal do espaço de desaceleração.


```python
v0 = 20.0  # m/s
a = -5.0   # m/s2
d_obstaculo = 50.0 # m

delta_s = -(v0**2) / (2 * a)
t_parada = -v0 / a
margem = d_obstaculo - delta_s

print(f'Velocidade Inicial:  {v0:.1f} m/s ({v0*3.6:.0f} km/h)')
print(f'Distância de Parada: {delta_s:.1f} m')
print(f'Tempo até Parada:    {t_parada:.1f} s')
print(f'Margem de Segurança: {margem:.1f} m (Colisão evitada: {margem > 0})')

```

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 4.0, 100)
s = v0 * t + 0.5 * a * t**2
v = v0 + a * t

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

ax1.plot(t, s, 'b-', lw=2.5, label='Posição s(t)')
ax1.axhline(50, color='r', linestyle='--', label='Obstáculo (50 m)')
ax1.scatter([4.0], [40.0], color='green', s=90, zorder=5, label='Parada Total (40 m)')
ax1.set_ylabel('Posição (m)', fontweight='bold')
ax1.set_title('Cinemática de Desaceleração: Posição e Velocidade no Tempo', fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.5)

ax2.plot(t, v, 'r-', lw=2.5, label='Velocidade v(t)')
ax2.axhline(0, color='black', lw=0.8)
ax2.set_xlabel('Tempo (segundos)', fontweight='bold')
ax2.set_ylabel('Velocidade (m/s)', fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.5)

plt.tight_layout()
plt.show()

```
