# Checkpoint 5: Dinâmica de Bloco em Plano Inclinado com Atrito
Verificação de equilíbrio estático versus escorregamento com atrito cinético.


```python
import numpy as np

m = 20.0 # kg
theta = np.radians(30.0)
mu_e = 0.40
mu_c = 0.30
g = 9.81

peso = m * g
n_force = peso * np.cos(theta)
p_x = peso * np.sin(theta)
fat_est_max = mu_e * n_force

escorrega = p_x > fat_est_max
fat_cin = mu_c * n_force
a_bloco = (p_x - fat_cin) / m if escorrega else 0.0

print(f'Componente Motora Px:    {p_x:.2f} N')
print(f'Atrito Estático Máximo:  {fat_est_max:.2f} N')
print(f'O bloco escorrega?       {escorrega}')
print(f'Atrito Cinético Atuante: {fat_cin:.2f} N')
print(f'Aceleração Rampa Abaixo: {a_bloco:.3f} m/s²')

```

```python
import matplotlib.pyplot as plt

forcas = ['Força Motora Px', 'Atrito Est. Máx', 'Atrito Cinético', 'Força Resultante']
valores = [p_x, fat_est_max, fat_cin, p_x - fat_cin]
cores = ['#d62728', '#ff7f0e', '#1f77b4', '#2ca02c']

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(forcas, valores, color=cores, width=0.5)
ax.set_ylabel('Força (Newtons)', fontweight='bold')
ax.set_title('Balanço Dinâmico de Forças no Plano Inclinado (θ = 30°)', fontweight='bold')
for b in bars:
    h = b.get_height()
    ax.text(b.get_x() + b.get_width()/2.0, h + 2, f'{h:.1f} N', ha='center', fontweight='bold')
ax.set_ylim(0, 120)
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
